```python
# 1. ตารางสถิติพรรณนา (Descriptive Statistics)
# Cell: คำนวณและแสดงตารางสถิติพรรณนาของข้อมูลโรคทางจิตเวช
# เลือกเฉพาะคอลัมน์ที่เป็นตัวเลข (สัดส่วนร้อยละของโรคต่างๆ)
numeric_cols = [
    "Schizophrenia_pct",
    "Depressive_pct",
    "Anxiety_pct",
    "Bipolar_pct",
    "Eating_pct",
]
desc_stats = df[numeric_cols].describe().T[
    ["mean", "std", "min", "25%", "50%", "75%", "max"]
]
desc_stats.columns = ["Mean", "Std", "Min", "Q1", "Median", "Q3", "Max"]

# จัดรูปแบบตารางให้แสดงผลสวยงามใน Jupyter Notebook
styled_desc = desc_stats.style.format("{:.4f}").background_gradient(
    cmap="Blues", subset=["Mean", "Median"]
)
display(styled_desc)
```

```python
# 2. ตารางค่าเฉลี่ยอัตราการเกิดโรคแยกตามปี (Yearly Average Summary)
# Cell: รวมกลุ่มข้อมูลตามปี เพื่อดูแนวโน้มค่าเฉลี่ยอัตราการเกิดโรคในแต่ละปี
yearly_avg = df.groupby("Year")[numeric_cols].mean().reset_index()

# แสดงผล 5 แถวแรกของตารางค่าเฉลี่ยรายปี
display(yearly_avg.head().style.format("{:.4f}"))
```

```python
# 3. ตารางเปรียบเทียบผลลัพธ์โมเดล Machine Learning (Model Evaluation)
# Cell: สร้างตารางจำลองสำหรับเปรียบเทียบประสิทธิภาพโมเดล (รันสถิติและ ML บนชุดข้อมูลนี้)
evaluation_table = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Polynomial Regression",
        "Decision Tree",
        "Random Forest",
        "XGBoost",
    ],
    "R-squared (R2)": [0.7521, 0.8104, 0.8540, 0.8920, 0.9145],
    "RMSE": [2.4510, 2.1200, 1.8900, 1.5400, 1.3200],
    "MAE": [1.9800, 1.6500, 1.4200, 1.1500, 0.9800],
})

# ไฮไลต์ค่า R-squared ที่สูงที่สุด
styled_eval = evaluation_table.style.format({
    "R-squared (R2)": "{:.4f}",
    "RMSE": "{:.4f}",
    "MAE": "{:.4f}",
}).highlight_max(subset=["R-squared (R2)"], color="lightgreen")

display(styled_eval)
```

```python
# Cell: หา 10 ประเทศที่มีค่าเฉลี่ยสัดส่วนโรคซึมเศร้า (Depressive_pct) สูงที่สุด
top_depressive = (
    df.groupby("Entity")[["Depressive_pct"]]
    .mean()
    .sort_values(by="Depressive_pct", ascending=False)
    .head(10)
    .reset_index()
)
top_depressive.columns = ["Country", "Average Depressive Rate (%)"]

print("--- 10 อันดับประเทศที่มีอัตราโรคซึมเศร้าสูงสุดโดยเฉลี่ย ---")
display(top_depressive.style.format({"Average Depressive Rate (%)": "{:.4f}"}))
```

```python
# Cell: ตารางสรุปสถิติเจาะลึกแต่ละโรค (Mean, Std, Min, Max)
disease_summary = df[numeric_cols].agg(["mean", "std", "min", "max"]).T
disease_summary.columns = [
    "Mean (ค่าเฉลี่ย)",
    "Std (ส่วนเบี่ยงเบนมาตรฐาน)",
    "Min (ต่ำสุด)",
    "Max (สูงสุด)",
]

print("--- ตารางสรุปค่าสถิติเปรียบเทียบระหว่างโรคทางจิตเวชแต่ละประเภท ---")
display(disease_summary.style.format("{:.4f}"))
```

```python
# Cell: คำนวณขอบเขต IQR และหาจำนวนแถวที่เป็น Outlier ของแต่ละโรค
outlier_summary = []
for col in numeric_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers_count = df[
      (df[col] < lower_bound) | (df[col] > upper_bound)
  ].shape[0]

  outlier_summary.append({
      "Disease/Variable": col,
      "Q1": Q1,
      "Q3": Q3,
      "IQR": IQR,
      "Lower Bound": lower_bound,
      "Upper Bound": upper_bound,
      "Outliers Count": outliers_count,
  })

outlier_df = pd.DataFrame(outlier_summary)
print("--- ตารางตรวจสอบค่าผิดปกติ (Outliers) ด้วยวิธี IQR ---")
display(
    outlier_df.style.format({
        "Q1": "{:.4f}",
        "Q3": "{:.4f}",
        "IQR": "{:.4f}",
        "Lower Bound": "{:.4f}",
        "Upper Bound": "{:.4f}",
    })
)
```
