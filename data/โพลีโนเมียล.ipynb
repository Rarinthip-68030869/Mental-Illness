```python
#แทนค่าว่างในรหัสประเทศด้วยunknow
df['Code'] = df['Code'].fillna('Unknow')
```

```python
#Polynomial Regression Model
#นำปีมาพยากรณ์สัดส่วนผู้ป่วยโรคซึมเศร้า
import numpy as np
import pandas as pd
import plotly.express as px

#กำหนดตัวแปร x(ปี),y(ซึมเศร้า)
data_sub = df[df['Entity'] == 'World'].dropna()

x = data_sub[['Year']].values
y = data_sub['Depressive disorders (share of population) - Sex: Both - Age: Age-standardized']

#วาดกราฟดูแนวโน้ม
fig = px.scatter(x=x.ravel(), y=y, labels={'x': 'Year', 'y': 'Depressive Share'})
fig.show()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

#เส้นตรง
lin = LinearRegression().fit(x, y)

#พหุนามดีกรี2
pf = PolynomialFeatures(degree=2, include_bias=False)
xp = pf.fit_transform(x)
q = LinearRegression().fit(xp, y)

xx = np.linspace(x.min(), x.max(), 200).reshape(-1, 1)
#วาดกราฟทำนายทับลงกราฟเดิม
fig = px.scatter(x=x.ravel(), y=y, labels={'x': 'Year', 'y': 'Depressive Share'})
fig.add_scatter(x=xx.ravel(), y=lin.predict(xx), mode='lines', name='Linear')
fig.add_scatter(x=xx.ravel(), y=q.predict(pf.transform(xx)), mode='lines', name='Degree 2')
fig.update_layout(template='plotly_white')
fig.show()
```

```python
from sklearn.metrics import r2_score

r2_lin = r2_score(y, lin.predict(x))
r2_poly = r2_score(y, q.predict(xp))

print("Poly Degree 2 R2:", round(r2_poly, 4))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline

#แบ่งข้อมูลเทรน70 เทส30
X_tr, X_te, y_tr, y_te = train_test_split(x, y, test_size=0.3, random_state=0)

#สร้างพิพไลน์
pip = make_pipeline(
    PolynomialFeatures(degree=2, include_bias=False),
    LinearRegression()
)
pip.fit(X_tr, y_tr)

print("Train R2:", round(pip.score(X_tr, y_tr), 4))
print("Test R2:", round(pip.score(X_te, y_te), 4))
```

```python
from sklearn.preprocessing import StandardScaler

for d in [1, 2, 3, 5]:
    p = make_pipeline(
        PolynomialFeatures(d, include_bias=False),
        StandardScaler(),
        LinearRegression()
    )
    p.fit(X_tr, y_tr)
    print(f'Degree {d} - Train R2: {round(p.score(X_tr, y_tr), 3)} - Test R2: {round(p.score(X_te, y_te), 3)}')
```
